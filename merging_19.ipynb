```python
import pandas as pd
import geopandas as gpd
from shapely import wkt
```

```python
ookla_df = pd.read_csv('F:\Economic2019\Ookla_Merged.csv')

ookla_df
```

```python
Viirs = pd.read_csv('F:\Economic2019\Viirs_Merged_Cleaned.csv')

Viirs
```

```python
# Convert Ookla geometry to GeoDataFrame
ookla_df['geometry'] = ookla_df['geometry'].apply(wkt.loads)
ookla_df = gpd.GeoDataFrame(ookla_df, geometry='geometry')

ookla_df
```

```python
# Convert VIIRS longitude/latitude to GeoDataFrame
Viirs = gpd.GeoDataFrame(
    Viirs, geometry=gpd.points_from_xy(Viirs.longitude, Viirs.latitude)
)

Viirs
```

```python
join_df = gpd.sjoin(Viirs, ookla_df, how='inner', predicate='intersects')
join_df.head()
```

```python
join_df.to_csv('Ookla_Viirs.csv', index=False)
```
