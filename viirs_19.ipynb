```python
invalid_corrdinates = df[(df['longitude'] < -180) | (df['longitude'] > 180) | (df['latitude'] < -90) | (df['latitude'] > 90)]
print(invalid_corrdinates)
```

```python
df['state'] = df['state'].str.strip().str.lower()
```

```python
df = df[df['light_intensity'] >= 0]
```

```python
df.shape
```

```python
Viirs = df.groupby(['state', 'longitude', 'latitude'], as_index=False).agg({
    'light_intensity': 'mean',
    'year': 'first',
    'month': 'count'
})
Viirs.rename(columns={'month': 'observations'}, inplace=True)
print(Viirs)
```

```python
Viirss = Viirs.copy()
Viirss['light_intensity'] = Viirss['light_intensity'] / Viirss['observations']
Viirs_avg = df.groupby(['longitude', 'latitude'], as_index=False).agg({
    'light_intensity': 'mean'
})
print(Viirs_avg.head())
```

```python
output_path = r"F:\Economic2019\Viirs_Merged_Cleaned.csv"
Viirs_avg.to_csv(output_path,index=False)
```
